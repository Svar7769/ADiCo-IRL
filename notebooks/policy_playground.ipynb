# Checkpoint test

```python
from policy.loader import load_policy
import torch

policy = load_policy(
    checkpoint_path = "checkpoints/navigation_3a3g_seed0.pt",
    obs_dim = 10,
    n_agents = 3
)

obs = torch.randn(3, 10)
actions = policy(obs)

print("Action Shape:", actions.shape)
print("action range:", actions.min().item(), "to", actions.max().item())
assert actions.shape == (3, 2)
assert actions.abs().max().item() <= 0.5 + 1e-5
print("Checkpoint load test passed.")
```

# Normal Sanity test

```python
# test 1: same obs should give same actions every time.
obs = torch.randn(3, 10)
a1 = policy(obs)
a2 = policy(obs)
assert torch.allclose(a1, a2)
print("Test 1 passed: same obs gives same actions.")

# Test 2: different obs should give different actions.
obs1 = torch.randn(3, 10)
obs2 = torch.randn(3, 10)
a1 = policy(obs1)
a2 = policy(obs2)
assert not torch.allclose(a1, a2)
print("Test 2 passed: different obs gives different actions.")

# Test 3: Different agent should produce different actions for the same obs
obs = torch.zeros(3, 10)
actions_same_obs = policy(obs)
print("Per agent action : \n", actions_same_obs)
```
